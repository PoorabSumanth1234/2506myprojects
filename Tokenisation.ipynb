```python
import nltk

# Download the required resource
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize

corpus = '''
Hello ! I am Poorab , an aspiring AI engineer and I want to have deeper insights about AI.
I want to replace Chatgpt
'''
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

```python
## TOkenizaion
## COnvert a paragraph to words and sentence to words
```

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
for sentence in documents:
  print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
##PUnctuations will also be treated as a specific word
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)

## full stop(.) is not a separate word here and for last full stop it will be considered as a separate word.
```
